# 01 · Explore the data
Uses the cleaned data from `2_prepare_data.py`.

```python
import os
import sys

os.chdir("..")              # work from the project folder, where data/ and src/ are
sys.path.insert(0, os.getcwd())

import matplotlib.pyplot as plt
import pandas as pd

from src.utils import load_config

cfg = load_config()
location = cfg["location"]

region = cfg["loc"]["ercot_region"]
load = pd.read_csv("data/interim/ercot_load.csv", index_col="timestamp", parse_dates=True)
weather = pd.read_csv(f"data/interim/weather_{location}.csv", index_col="timestamp", parse_dates=True)
df = weather.join(load[[region]], how="inner")
df.describe().T
```

## Daily load over the years

```python
df[region].resample("D").mean().plot(figsize=(12, 3), ylabel="MW", title=f"{region} daily average load")
plt.show()
```

## Average day, by month

```python
profile = df.groupby([df.index.month, df.index.hour])[region].mean().unstack(0)
profile.plot(figsize=(10, 4), xlabel="Hour of day", ylabel="MW", title="Average hourly load per month (one line per month)", legend=False)
plt.show()
```

## Load vs temperature

```python
sample = df.sample(5000, random_state=0)
plt.figure(figsize=(6, 4))
plt.scatter(sample["Temperature"], sample[region], s=4, alpha=0.4)
plt.xlabel("Temperature (F)")
plt.ylabel("MW")
plt.title("Load vs temperature")
plt.show()
```
